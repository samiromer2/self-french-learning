# 02 — Data cleaning

Checks that belong in a junior-analyst portfolio: missing values, invalid geos, percent bounds, duplicate keys.

```python
from pathlib import Path
import pandas as pd

ROOT = Path("..") if Path("../data/processed/official_languages_long.csv").exists() else Path(".")
long = pd.read_csv(ROOT / "data/processed/official_languages_long.csv")
long.head()
```

```python
print(long.isna().sum())
print("duplicates", long.duplicated().sum())
print("negative values", (long.value < 0).sum())
pct = long[long.measure == "percent"]
print("percent > 100", (pct.value > 100.0001).sum())
print("geo_code values", sorted(long.geo_code.unique()))
print("years", sorted(long.year.unique()))
```

Mother tongue in this extract stops before 2021 even though the table title mentions 2021. That is a source limitation, not a pipeline bug — label the year instead of filling 2021 from 2016.
