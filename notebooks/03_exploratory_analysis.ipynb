# 03 — Exploratory analysis

Distributions and trends from processed facts. Plots use Matplotlib.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path("..") if Path("../data/processed/fact_knowledge_official_languages.csv").exists() else Path(".")
k = pd.read_csv(ROOT / "data/processed/fact_knowledge_official_languages.csv")
k.describe()[["bilingual_rate", "total"]]
```

```python
ca = k[k.geo_code == "CA"].sort_values("year")
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(ca.year, ca.bilingual_rate, color="#171717")
ax.set_ylabel("English–French bilingualism (%)")
ax.set_xlabel("Census year")
ax.set_title("Canada")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
ca[["year", "bilingual_rate", "english_and_french", "total"]]
```

```python
latest = k.year.max()
prov = k[(k.year == latest) & (k.geo_type == "province")].sort_values("bilingual_rate")
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(prov.geo_name, prov.bilingual_rate, color="#444")
ax.set_xlabel(f"Bilingualism rate (%) · {latest}")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```
