# 04 — French in Canada

Answers the dashboard questions with pandas. Personal learning data is intentionally absent.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path("..") if Path("../data/analytics/dashboard_bundle.json").exists() else Path(".")
bundle = json.loads((ROOT / "data/analytics/dashboard_bundle.json").read_text())
k = pd.read_csv(ROOT / "data/processed/fact_knowledge_official_languages.csv")
home = pd.read_csv(ROOT / "data/processed/fact_home_language.csv")
work = pd.read_csv(ROOT / "data/processed/fact_language_at_work.csv")
imm = pd.read_csv(ROOT / "data/processed/fact_immigrant_official_languages.csv")
```

## Q1–Q2 Bilingualism over time and by province

```python
ca = k[k.geo_code == "CA"].sort_values("year")
print(ca[["year", "bilingual_rate"]].to_string(index=False))
print("change 1951→latest pp", float(ca.iloc[-1].bilingual_rate - ca.iloc[0].bilingual_rate))
latest = k.year.max()
print(k[(k.year == latest) & (k.geo_type == "province")][["geo_name", "bilingual_rate"]].sort_values("bilingual_rate", ascending=False).to_string(index=False))
```

## Q3–Q4 French at home vs at work

```python
h = home[home.year == home.year.max()][["geo_code", "geo_name", "french_rate"]].rename(columns={"french_rate": "home"})
w = work[["geo_code", "french_rate"]].rename(columns={"french_rate": "work"})
cmp = h.merge(w, on="geo_code", how="left")
print(cmp.sort_values("home", ascending=False).to_string(index=False))
print("These universes differ: all people vs workers aged 15+ who worked since 2020-01-01.")
```

## Q5–Q6 Geography and immigrant status

```python
print(k[(k.year == latest) & (k.geo_code.isin(["QC", "CA-XQ", "ON", "NB"]))][["geo_name", "bilingual_rate"]])
i2021 = imm[(imm.year == latest) & (imm.geo_code.isin(["CA", "QC", "CA-XQ"])) & (imm.immigrant_status_key.isin(["immigrants", "non_immigrants"]))]
print(i2021[["geo_name", "immigrant_status_key", "french_knowledge_rate", "bilingual_rate"]].to_string(index=False))
```
